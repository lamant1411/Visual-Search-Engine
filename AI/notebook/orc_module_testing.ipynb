```python
# Hệ thống sẽ khởi tạo Reader và load cấu hình ngôn ngữ Anh-Việt
extractor = OCRExtractor()
```

```python
import os
import sys


CURRENT_DIR = os.getcwd()
PARENT_DIR = os.path.dirname(CURRENT_DIR)


if PARENT_DIR not in sys.path:
    sys.path.append(PARENT_DIR)


test_image_path = os.path.join(PARENT_DIR, "data", "sample_images", "imagesign.jpg")

if os.path.exists(test_image_path):
    print(f"Tìm thấy file ảnh thử nghiệm tại: {test_image_path}")
else:
    print(f"Cảnh báo: Không tìm thấy ảnh tại {test_image_path}. Vui lòng kiểm tra lại!")
```

```python
if test_image_path and os.path.exists(test_image_path):
    print("--- Bắt đầu chạy trích xuất OCR ---")
    extracted_texts = extractor.extract_text(test_image_path)
    print(f"Thành công! Tìm thấy {len(extracted_texts)} dòng chữ.")
    print(f"Kết quả đầu ra (list[str]): {extracted_texts}")
else:
    print("Vui lòng cập nhật đường dẫn ảnh hợp lệ trước khi thực hiện cell này.")
```
