```python
import os
import easyocr
import pandas as pd
from tqdm import tqdm

# Cấu hình đường dẫn
IMAGE_DIR = "/content/open_images"
# Lưu thẳng file kết quả lên Drive để không bị mất khi Colab sập
OUTPUT_CSV = "/content/drive/MyDrive/ocr_metadata.csv"
SAVE_INTERVAL = 100

print("Đang khởi tạo EasyOCR với GPU...")
reader = easyocr.Reader(['en'], gpu=True)

image_files = [f for f in os.listdir(IMAGE_DIR) if f.endswith(".jpg")]
print(f"Tổng số ảnh trong thư mục: {len(image_files)}")

processed_ids = set()
if os.path.exists(OUTPUT_CSV):
    existing_df = pd.read_csv(OUTPUT_CSV)
    processed_ids = set(existing_df['image_id'].tolist())
    print(f"Đã có {len(processed_ids)} ảnh được xử lý từ lần trước. Chạy tiếp phần còn lại...")

results = []
new_count = 0

for filename in tqdm(image_files, desc="Trích xuất OCR"):
    image_id = filename.split(".")[0]

    if image_id in processed_ids:
        continue

    image_path = os.path.join(IMAGE_DIR, filename)

    try:
        extracted_texts = reader.readtext(image_path, detail=0)
        full_text = " ".join(extracted_texts).strip()

        results.append({
            "image_id": image_id,
            "ocr_text": full_text
        })
        new_count += 1

        if new_count % SAVE_INTERVAL == 0:
            pd.DataFrame(results).to_csv(OUTPUT_CSV, mode='a', header=not os.path.exists(OUTPUT_CSV), index=False)
            results = []

    except Exception as e:
        pass # Bỏ qua âm thầm nếu ảnh bị lỗi để không dừng vòng lặp

# Lưu phần còn lại
if results:
    pd.DataFrame(results).to_csv(OUTPUT_CSV, mode='a', header=not os.path.exists(OUTPUT_CSV), index=False)

print("\nHoàn tất 100%!")
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, lit, concat_ws, collect_list, lower, regexp_replace, trim, concat
import os

# 1. Khởi tạo Spark Session
spark = SparkSession.builder \
    .appName("VisualSearch_Data_Pipeline") \
    .getOrCreate()


# PHẦN 1: DỮ LIỆU UNSPLASH (25,000 ảnh)
# Đọc file
df_uns_photos = spark.read.csv("/content/drive/MyDrive/Unsplash_Lite/photos.tsv000", sep='\t', header=True)
df_uns_keywords = spark.read.csv("/content/drive/MyDrive/Unsplash_Lite/keywords.tsv000", sep='\t', header=True)

# Gộp keywords
df_keywords_grouped = df_uns_keywords.groupby("photo_id") \
    .agg(concat_ws(", ", collect_list("keyword")).alias("all_keywords"))

# Join và map cột
df_unsplash_clean = df_uns_photos.join(df_keywords_grouped, on="photo_id", how="left") \
    .select(
        col("photo_id").alias("image_id"),
        col("photo_image_url").alias("image_path"),
        lit("unsplash").alias("source"),
        concat_ws(" - ", col("ai_description"), col("all_keywords")).alias("semantic_text")
    )


# PHẦN 2: DỮ LIỆU OPEN IMAGES (60,000 ảnh)
# 1. LẤY DANH SÁCH ẢNH THỰC TẾ TRONG Ổ CỨNG
image_dir = "/content/open_images"
try:
    valid_image_ids = [f.split(".")[0] for f in os.listdir(image_dir) if f.endswith(".jpg")]
    print(f"Đã quét được {len(valid_image_ids)} ảnh Open Images thực tế.")
except FileNotFoundError:
    print(f"Lỗi: Không tìm thấy thư mục {image_dir}. Hãy giải nén ảnh trước!")
    valid_image_ids = []

# Tạo DataFrame chứa ID hợp lệ để lọc
df_valid_ids = spark.createDataFrame([(i,) for i in valid_image_ids], ["ImageID"])

# 2. ĐỌC FILE METADATA GỐC
df_oi_raw = spark.read.csv("/content/drive/MyDrive/open_images_metadata_backup/classifications.csv", header=True)

# 3. LỌC: Giữ lại những dòng có ID nằm trong ổ cứng (Inner Join)
df_oi_filtered = df_oi_raw.join(df_valid_ids, on="ImageID", how="inner")

# 4. GOM NHÓM: Ghép tất cả LabelName của cùng 1 ảnh thành 1 chuỗi (Giống Unsplash)
df_oi_grouped = df_oi_filtered.groupBy("ImageID") \
    .agg(concat_ws(", ", collect_list("LabelName")).alias("semantic_text"))

# 5. ĐỔI TÊN CỘT VÀ TẠO PATH
base_path = "open_images/"

df_oi_clean = df_oi_grouped.select(
    col("ImageID").alias("image_id"),
    concat(lit(base_path), col("ImageID"), lit(".jpg")).alias("image_path"),
    lit("open_images").alias("source"),
    lower(col("semantic_text")).alias("semantic_text")
)


# ==========================================
# PHẦN 3: GỘP, CHUẨN HÓA TEXT & XUẤT FILE
# ==========================================
# 1. Gộp 2 DataFrame (Unsplash + Open Images)
df_final_unified = df_unsplash_clean.unionByName(df_oi_clean, allowMissingColumns=True)

# 2. ĐỌC DỮ LIỆU OCR TỪ FILE ĐÃ CHẠY TRƯỚC ĐÓ
try:
    df_ocr = spark.read.csv("/content/drive/MyDrive/ocr_metadata.csv", header=True)
    # JOIN cột ocr_text vào bảng chính
    df_final_unified = df_final_unified.join(df_ocr, on="image_id", how="left")
    print("Đã tích hợp thành công dữ liệu OCR!")
except Exception as e:
    print("Chưa có file OCR. Hệ thống sẽ tạo cột ocr_text trống (NULL).")
    df_final_unified = df_final_unified.withColumn("ocr_text", lit(None))

# 3. HÀM CHUẨN HÓA TEXT
def normalize_text(col_name):
    clean_col = regexp_replace(col_name, r"[^a-zA-Z0-9\s]", "")
    return trim(regexp_replace(clean_col, r"\s+", " "))

# Chuẩn hóa cả 2 cột text
df_final_unified = df_final_unified.withColumn("semantic_text", normalize_text(col("semantic_text")))
df_final_unified = df_final_unified.withColumn("ocr_text", normalize_text(col("ocr_text")))

# 4. XUẤT FILE: Dùng coalesce(1) để gom về 1 file CSV duy nhất
output_folder = "/content/drive/MyDrive/visual_search_unified_database_folder"

df_final_unified.coalesce(1).write.csv(
    output_folder,
    header=True,
    mode="overwrite"
)

print(f"Pipeline hoàn tất! Dữ liệu đã lưu tại thư mục: {output_folder}")
```

```python
import pandas as pd

# 1. Cấu hình đường dẫn file
INPUT_CSV = "/content/drive/MyDrive/visual_search_unified_database.csv"
CLEANED_CSV = "/content/drive/MyDrive/cleaned_metadata_85k.csv"

print("Đang tải dữ liệu...")
df = pd.read_csv(INPUT_CSV)
print(f"Số lượng dòng ban đầu: {len(df)}")

# 2. Làm sạch dữ liệu
print("Đang tiến hành làm sạch...")

# Thay thế các giá trị NULL/NaN bằng chuỗi rỗng ("") trong các cột chứa text
df['ocr_text'] = df['ocr_text'].fillna("")
df['semantic_text'] = df['semantic_text'].fillna("")

# Ép kiểu toàn bộ 2 cột này về string
df['ocr_text'] = df['ocr_text'].astype(str)
df['semantic_text'] = df['semantic_text'].astype(str)

# (Tùy chọn) Loại bỏ các khoảng trắng thừa ở đầu/cuối chuỗi
df['ocr_text'] = df['ocr_text'].str.strip()
df['semantic_text'] = df['semantic_text'].str.strip()

# Loại bỏ các dòng bị trùng lặp (nếu có) dựa trên image_id
df = df.drop_duplicates(subset=['image_id'], keep='first')

# 3. Kiểm tra lại và lưu trữ
print(f"Số lượng dòng sau khi làm sạch: {len(df)}")

# Kiểm tra xem còn giá trị null nào trong cột text không
null_ocr_count = df['ocr_text'].isna().sum()
print(f"Số lượng giá trị NULL trong ocr_text hiện tại: {null_ocr_count}")

# Lưu file dữ liệu đã làm sạch
df.to_csv(CLEANED_CSV, index=False, encoding='utf-8')
print(f"Hoàn tất! Đã lưu file dữ liệu sạch tại: {CLEANED_CSV}")
```
